```python
import numpy as np
import pandas as pd
import yfinance as yf
import requests
```

```python
symbol = "NSE_INDEX%7CNifty%2050"

# nifty 50 historical data params
interval = "days"
frequency = "1"
to_date = "2025-12-16"
from_date = "2025-11-11"
```

```python
# for nifty 50 historical data
url = f"https://api.upstox.com/v3/historical-candle/{symbol}/{interval}/{frequency}/{to_date}/{from_date}"
response = requests.get(url)
if response.status_code == 200:
    print(response.json())
else:
    print(f"Error: {response.status_code} - {response.text}")
```

```python
nifty_df = pd.DataFrame()
nifty_df['Date'] = pd.to_datetime([item[0].split("T")[0] for item in response.json()['data']['candles']])
nifty_df['Close'] = [item[4] for item in response.json()['data']['candles']]
nifty_df['Open'] = [item[1] for item in response.json()['data']['candles']]
nifty_df['High'] = [item[2] for item in response.json()['data']['candles']]
nifty_df['Low'] = [item[3] for item in response.json()['data']['candles']]
nifty_df.set_index('Date', inplace=True)
nifty_df.sort_index(inplace=True)
nifty_df['logReturn'] = np.log(nifty_df['Close']/nifty_df['Close'].shift(1))
nifty_df['pctReturn'] = nifty_df['Close'].pct_change()
nifty_df
```

```python
nifty_df.to_csv('data/nifty_50_historical_data.csv')
```
